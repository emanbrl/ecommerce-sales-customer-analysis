### 7.2 Unit Price

The `UnitPrice` variable contains zero, negative, and extreme values.
These observations need to be investigated to determine whether they
represent valid business transactions, special cases, or data quality
issues.

```python
df["UnitPrice"].describe()
```

```python
print("Zero prices:", (df["UnitPrice"] == 0).sum())
print("Negative prices:", (df["UnitPrice"] < 0).sum())
print("Prices above 100:", (df["UnitPrice"] > 100).sum())
```
