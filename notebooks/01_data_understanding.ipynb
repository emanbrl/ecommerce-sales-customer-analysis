## 8. Initial Findings

Based on the initial investigation, we identified several areas
requiring further investigation before data cleaning:

### 8.1 Business Entities

- The dataset contains 541,909 transaction lines, representing 25,900 unique invoices.

- There are 4,372 unique identified customers and 4,070 unique products across 38 countries.

- The number of transaction lines is much larger than the number of invoices because a single invoice can contain multiple product lines.

- CustomerID is missing for 135,080 transaction lines, which means customer-level analysis may require careful treatment of transactions without an identified customer.

- The high frequency of some InvoiceNo and StockCode values also confirms that the dataset is structured at the transaction-line level rather than at the order or customer level.

### 8.2 Negative Quantities

- The dataset contains 10,624 transaction lines with negative quantities.

- Among these, 9,288 rows have an `InvoiceNo` beginning with `C`. All 9,288 C-prefixed invoice rows have negative quantities, indicating a strong relationship between the `C` prefix and negative-quantity transactions.

- However, 1,336 negative-quantity rows do not have a C prefix. Therefore, the relationship is not one-to-one.

- At this stage, negative quantities should not be removed automatically. Their business meaning and appropriate treatment will be determined during the data-cleaning stage based on the project's analytical objectives.

### 8.3 Unit Price

- 2,515 rows have UnitPrice = 0
- 2 rows have negative UnitPrice, both "Adjust bad debt"
- 1,036 rows have UnitPrice > 100, but these include both expensive products and special transactions such as postage, fees, discounts, and manual entries
- High UnitPrice values cannot be classified as errors simply because they are high. They include legitimate expensive products as well as postage, fees, discounts, and manual transactions.
